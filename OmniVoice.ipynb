```python
input_dir = '/content/data/audio'  # Chemin de votre répertoire audio
temp_dir = './temp_audio'

# Créer un dossier temporaire
os.makedirs(temp_dir, exist_ok=True)

for filename in os.listdir(input_dir):
    if filename.endswith('.wav'):
        input_path = os.path.join(input_dir, filename)
        temp_path = os.path.join(temp_dir, filename)

        # 1. Convertir dans le dossier temporaire avec le même nom
        !ffmpeg -y -i "{input_path}" -ar 24000 "{temp_path}"

        # 2. Remplacer l'original par le fichier converti
        os.replace(temp_path, input_path)

# Nettoyer le dossier temporaire
os.rmdir(temp_dir)

print("Conversion et remplacement terminés avec le même nom !")
```

### 3.5 Evaluation

Evaluation BASSA

```python
model_bassa = OmniVoice.from_pretrained(
    "drive/MyDrive/Omnivoice/bassa/20260919/checkpoint-750",
    device_map="cuda:0")
```

```python
audios = model_bassa.generate(
    text="I ngwélél kunde yosôna i nuga bikai bisu, nuga bikai bisu i béñgege. I mboñ ki le hisi lôñni bôt ba nyééne mu ba bégés nuga bikai bisu; nuga bikai bisu i bééna kwéé i i bé le i nol yo, ndi kwéé yé i bi mbôôp. Bibégés bi ba ni Nyambe Tata nu Nwet wés Yésu Kristô inyule a bi ti bés ligwéé li yondo inoñnaga ni ngandak yé konangoo le ndi di bana botñem i i nset bé ni njel bitugne bi Yésu Kristô ikété bawoga",
    ref_audio="drive/MyDrive/Omnivoice/bassa/data/Sakaria_38_8.wav",
    ref_text="Me bi tehe yiinda juu. Mut wada a bé a yii i ngii kôyôp hosi a bé a téé ipôla bie bi myrte nyoo isi hikôa; bakilhosi ba bé i mbus yé ba bak i ngii bihosi bi bi nkôyôp ni bihosi bi bi yé minik minik lôñni bihosi bi bi yé puba."
)
display(Audio(audios[0], rate=24000))
```
