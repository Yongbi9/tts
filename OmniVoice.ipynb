## 3. Option B — Python API

### 3.1 Load Model

```python
from omnivoice import OmniVoice
import soundfile as sf
import torch
from IPython.display import Audio, display

model = OmniVoice.from_pretrained(
    "k2-fsa/OmniVoice",
    device_map="cuda:0",
    dtype=torch.float16,
    load_asr=True,
)
```

### 3.2 Voice Cloning

Clone a voice from a short (3-10s) reference audio clip. Upload your own `ref.wav` or use any audio file.

`ref_text` is optional — if omitted, the model uses Whisper ASR to auto-transcribe it.

```python
from google.colab import files

print("Upload a reference audio file (wav/mp3/flac):")
uploaded = files.upload()
ref_audio_path = list(uploaded.keys())[0]
print(f"Uploaded: {ref_audio_path}")
```

```python
!ffmpeg -i Sakaria_38_8.wav -ar 24000 Sakaria_38_8_24k.wav
```

```python
ref_audio_path = "Sakaria_38_8_24k.wav"
```

```python
audio = model.generate(
    text="Mè yega",
    ref_audio=ref_audio_path,
    ref_text="Me bi tehe yiinda juu. Mut wada a bé a yii i ngii kôyôp hosi a bé a téé ipôla bie bi myrte nyoo isi hikôa; bakilhosi ba bé i mbus yé ba bak i ngii bihosi bi bi nkôyôp ni bihosi bi bi yé minik minik lôñni bihosi bi bi yé puba.",  # optional
)

sf.write("clone_out.wav", audio[0], 24000)
display(Audio(audio[0], rate=24000))
```

### 3.3 Reusing a cloned voice across sessions

Encode the reference audio once, save the resulting prompt, and skip the audio loading / auto-transcription steps in later sessions\

```python
prompt = model.create_voice_clone_prompt(
    ref_audio=ref_audio_path,
    ref_text="Me bi tehe yiinda juu. Mut wada a bé a yii i ngii kôyôp hosi a bé a téé ipôla bie bi myrte nyoo isi hikôa; bakilhosi ba bé i mbus yé ba bak i ngii bihosi bi bi nkôyôp ni bihosi bi bi yé minik minik lôñni bihosi bi bi yé puba.",  # optional
)
prompt.save("my_voice.pt")

# Later, in a new session:
from omnivoice import VoiceClonePrompt
```

```python
prompt = VoiceClonePrompt.load("my_voice.pt")
audio = model.generate(
    text="Yéhôva a kal Satan le: I gwom gwobisôna a gwé bi yé i moo moñ. Ndi nyemede u tihba bañ nye!",
    voice_clone_prompt=prompt)
display(Audio(audio[0], rate=24000))
```
